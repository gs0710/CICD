**Create Sales Dataset**

```python
import findspark
findspark.init()
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SalesData").getOrCreate()

sales_data = [
    (101, "Laptop", 60000, 1),
    (102, "Mobile", 20000, 2),
    (103, "Shoes", 3000, 3),
    (104, "Tablet", 25000, 1),
    (105, "Headphones", 4000, 4)
]

sales = spark.createDataFrame(sales_data, schema=["OrderID", "Product", "Price", "Quantity"])

sales.show()
```

**Select Specific Columns**

```python
sales.select("Product","Price").show()
```

**Filter Products (Price > 10000)**

```python
from pyspark.sql.functions import col
sales.filter(col("Price") > 10000).show()
```

**Exercise 2: Movie Dataset Analysis**
Create Movie Dataset or Define Movie Schema and Data.

```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, IntegerType

movies_data = [
    ("Avatar","Sci-Fi",9.1,2800),
    ("Titanic","Drama",8.5,2200),
    ("Avengers","Action",9.0,2600),
    ("Inception","Sci-Fi",8.8,830),
    ("Joker","Drama",8.7,1000)
]

movies_schema = StructType([
    StructField("Movie", StringType(), True),
    StructField("Genre", StringType(), True),
    StructField("Rating", DoubleType(), True),
    StructField("Revenue", IntegerType(), True)
])

movies = spark.createDataFrame(movies_data, schema=movies_schema)

movies.show()
```

**Create SQL Table**

```python
movies.createOrReplaceTempView("movies")
```

**Display All Movies**

```python
spark.sql("SELECT * FROM movies").show()
```

**Highest Rated Movie**

```python
spark.sql("""
SELECT Movie, Rating
FROM movies
ORDER BY Rating DESC
LIMIT 1
""").show()
```

**Count Movies by Genre**

```python
spark.sql("""
SELECT Genre, COUNT(*) AS TotalMovies
FROM movies
GROUP BY Genre
""").show()
```

**StudentTasks**
Average Movie Rating

```python
spark.sql("""
SELECT AVG(Rating) AS AverageRating
FROM movies
""").show()
```

**Movie with Highest Revenue**

```python
spark.sql("""
SELECT Movie, Revenue
FROM movies
ORDER BY Revenue DESC
LIMIT 1
""").show()
```

**Count Movies in Each Genre**

```python
spark.sql("""
SELECT Genre, COUNT(*) AS Count
FROM movies
GROUP BY Genre
""").show()
```
